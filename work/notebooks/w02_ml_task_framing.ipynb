## 1. My lane as an ML task (type)

This project is an object detection task. The model detects two objects in an image: a mobile phone and a seatbelt. Based on these detections, it identifies whether a driver is following safety rules.

```python
task_type = "Object Detection"
print(task_type)
```

## 2. Target or proxy

The target is to detect the presence of a mobile phone and whether the driver is wearing a seatbelt. The labels come from a publicly annotated Kaggle dataset where each image is manually labeled.

```python
target_classes = ["mobile_phone", "seatbelt"]
print(target_classes)
```

## 3. Success metric

The primary evaluation metric is Mean Average Precision (mAP). A higher mAP means the model detects objects more accurately. My project achieved approximately 88% mAP.

```python
map_score = 0.88
print(f"mAP: {map_score}")
```

## 4. The unit of analysis, as a real dataframe

One row represents one labeled image from the dataset. Each image contains annotations such as image ID, bounding box coordinates, object class (phone or seatbelt), and other metadata used for training.

```python
import pandas as pd

df = pd.DataFrame({
    "image_name": ["1.jpg", "2.jpg", "3.jpg", "4.jpg", "5.jpg"],
    "phone_detected": [1, 0, 1, 0, 1],
    "seatbelt_detected": [1, 1, 0, 1, 0]
})

display(df)
print("Total sample images:", len(df))
```

## 5. Why ML beats a fixed rule here

A fixed rule cannot reliably detect phones or seatbelts because lighting, camera angles, driver positions, and object appearances vary significantly. Machine learning learns these complex visual patterns from many labeled examples and generalizes to new images.

```python
reason = "ML handles varying lighting, camera angles, and object positions better than fixed rules."
print(reason)
```
