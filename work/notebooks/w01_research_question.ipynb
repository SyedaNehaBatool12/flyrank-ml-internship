## Initial observations

The dataset contains 30,000 pages with 44 features.

After filtering pages with impressions greater than zero, the correlation between search volume and impressions remained extremely low (0.001). This suggests that search volume alone is not a reliable indicator of page traffic in this dataset.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

filtered = df[df["impressions_90d"] > 0]
corr = filtered["search_volume"].corr(filtered["impressions_90d"])

print("Correlation:", round(corr, 3))
```
